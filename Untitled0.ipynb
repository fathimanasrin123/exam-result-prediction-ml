```python
import pandas as pd

data = pd.read_csv(
    "student_data/extracted/student-mat.csv",
    sep=';'
)

print("Dataset loaded successfully!")
print("Number of rows:", data.shape[0])
print("Number of columns:", data.shape[1])

data.head()
```

```python
# Create Pass/Fail result

data['Result'] = data['G3'].apply(lambda x: 1 if x >= 10 else 0)

# 1 = Pass, 0 = Fail

print(data[['G3', 'Result']].head(10))

print("\nResult count:")
print(data['Result'].value_counts())
```

```python
# Check dataset information

print("Dataset Information:")
data.info()

print("\nMissing values:")
print(data.isnull().sum())

print("\nDuplicate rows:")
print(data.duplicated().sum())
```

```python
# Remove duplicate rows

data = data.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(data.shape)
```

```python
# Select input features and target

features = ['studytime', 'failures', 'absences', 'G1', 'G2']

X = data[features]
y = data['Result']

print("Input features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
# Check for missing values in selected features

print("Missing values in features:")
print(X.isnull().sum())

print("\nFeature statistics:")
print(X.describe())
```

```python
import matplotlib.pyplot as plt

result_counts = data['Result'].value_counts()

plt.figure(figsize=(6,4))

plt.bar(
    ['Fail', 'Pass'],
    [result_counts.get(0, 0), result_counts.get(1, 0)]
)

plt.xlabel('Exam Result')
plt.ylabel('Number of Students')
plt.title('Pass vs Fail Students')

plt.show()
```

```python
plt.figure(figsize=(7,5))

data.groupby('studytime')['Result'].mean().plot(kind='bar')

plt.xlabel('Study Time')
plt.ylabel('Pass Rate')
plt.title('Study Time vs Exam Result')

plt.xticks(rotation=0)
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

```python
from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(random_state=42)

# Train model
model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict results for test data

y_pred = model.predict(X_test_scaled)

print("Predicted results:")
print(y_pred)

print("\nActual results:")
print(y_test.values)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=['Fail', 'Pass']
))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Fail', 'Pass']
)

disp.plot()

plt.title('Confusion Matrix')
plt.show()
```

```python
metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

values = [accuracy, precision, recall, f1]

plt.figure(figsize=(8,5))

plt.bar(metrics, values)

plt.ylim(0, 1)

plt.xlabel('Evaluation Metrics')
plt.ylabel('Score')
plt.title('Performance of Exam Result Prediction Model')

plt.show()
```

```python
comparison = pd.DataFrame({
    'Actual Result': y_test.values,
    'Predicted Result': y_pred
})

print(comparison.head(20))
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    range(len(y_test)),
    y_test.values,
    marker='o',
    label='Actual'
)

plt.plot(
    range(len(y_pred)),
    y_pred,
    marker='x',
    label='Predicted'
)

plt.xlabel('Test Student')
plt.ylabel('Result')
plt.title('Actual vs Predicted Exam Results')

plt.yticks([0, 1], ['Fail', 'Pass'])

plt.legend()
plt.show()
```

```python
# Example new student

new_student = [[3, 0, 5, 12, 13]]

# Scale the new student's data
new_student_scaled = scaler.transform(new_student)

# Make prediction
prediction = model.predict(new_student_scaled)

if prediction[0] == 1:
    print("Predicted Exam Result: PASS")
else:
    print("Predicted Exam Result: FAIL")
```
